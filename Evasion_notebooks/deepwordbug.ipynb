```python
# STEP 2: Import libraries
from textattack.attack_recipes import DeepWordBugGao2018
from textattack.models.wrappers import HuggingFaceModelWrapper
from transformers import AutoModelForSequenceClassification, AutoTokenizer
from textattack.datasets import Dataset
from textattack import Attacker, AttackArgs
import pandas as pd
```

```python
# STEP 3: Load Enron dataset
df = pd.read_csv("enron1_train.csv")
spam_texts = df[df["target"] == "spam"]["email"].head(1000).tolist()
examples = [(text, 1) for text in spam_texts]  # Label 1 = spam
dataset = Dataset(examples)
```

```python
# STEP 4: Load classifier (pretrained spam model)
model_name = "mrm8488/bert-tiny-finetuned-sms-spam-detection"
model = AutoModelForSequenceClassification.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)
model_wrapper = HuggingFaceModelWrapper(model, tokenizer)
```

```python
# STEP 5: Load DeepWordBug attack
attack = DeepWordBugGao2018.build(model_wrapper)
```

```python
# STEP 6: Configure attack runner
attack_args = AttackArgs(
    num_examples=len(examples),
    disable_stdout=False,
    log_to_txt="deepwordbug_results.txt",  # Optional: saves results to file
)
```

```python
# STEP 7: Run attack
attacker = Attacker(attack, dataset, attack_args)
attacker.attack_dataset()
```
